Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
population = pd.read_csv("WB_WDI_SP_POP_TOTL.csv")
gender = pd.read_csv("WB_GS_SL_UEM_1524_FM_ZS.csv")
```

Data Preparation (South African population)

```python
population_sa = population[
    population["REF_AREA"] == "ZAF"
].copy()
```

```python
print(population_sa.shape)
```

```python
print(population_sa[[
    "REF_AREA",
    "REF_AREA_LABEL",
    "TIME_PERIOD",
    "OBS_VALUE",
    "INDICATOR_LABEL"
]].head(10))
```

```python
print(population_sa["TIME_PERIOD"].min(),
      population_sa["TIME_PERIOD"].max())

print("Number of years:",
      population_sa["TIME_PERIOD"].nunique())

print("Duplicate years:",
      population_sa["TIME_PERIOD"].duplicated().sum())
```

Cleaning 

```python
population_clean = population_sa[
    (population_sa["TIME_PERIOD"] >= 1991) &
    (population_sa["TIME_PERIOD"] <= 2023)
].copy()
```

```python
population_clean = population_clean[
    ["TIME_PERIOD", "OBS_VALUE"]
]
```

```python
population_clean = population_clean.rename(columns={
    "TIME_PERIOD": "Year",
    "OBS_VALUE": "Population"
})
```

```python
population_clean = population_clean.sort_values("Year")
print(population_clean.head())
print(population_clean.tail())
```

```python
print("Rows:", len(population_clean))
print("Missing values:")
print(population_clean.isnull().sum())
```

Data Preparation (Gender dataset)

```python
print(gender[gender["REF_AREA"] == "ZAF"]["REF_AREA_LABEL"].unique())
print(gender[gender["REF_AREA"] == "ZAF"]["INDICATOR_LABEL"].unique())
print(gender[gender["REF_AREA"] == "ZAF"][
    ["TIME_PERIOD", "OBS_VALUE", "COMP_BREAKDOWN_1", "COMP_BREAKDOWN_1_LABEL",
     "COMP_BREAKDOWN_2", "COMP_BREAKDOWN_2_LABEL",
     "COMP_BREAKDOWN_3", "COMP_BREAKDOWN_3_LABEL"]
].drop_duplicates().head(30))
```

```python
gender_modeled = gender[
    (gender["REF_AREA"] == "ZAF") &
    (gender["COMP_BREAKDOWN_1"] == "GS_TCZ") &
    (gender["TIME_PERIOD"] >= 1991) &
    (gender["TIME_PERIOD"] <= 2023)
].copy()
print(gender_modeled.shape)
```

```python
print(gender_modeled[
    ["TIME_PERIOD", "OBS_VALUE", "COMP_BREAKDOWN_1_LABEL"]
].to_string(index=False))
print(gender_modeled["OBS_VALUE"].isna().sum())
```

```python
print(gender_modeled["OBS_VALUE"].notna().sum())
```

```python
print("Duplicate years:", gender_modeled["TIME_PERIOD"].duplicated().sum())
```

Cleaning

```python
gender_clean = gender_modeled[
    ["TIME_PERIOD", "OBS_VALUE"]
].copy()
gender_clean = gender_clean.rename(columns={
    "TIME_PERIOD": "Year",
    "OBS_VALUE": "Female_Male_Youth_Unemployment_Ratio"
})
gender_clean = gender_clean.sort_values("Year")
print(gender_clean.head())
print(gender_clean.tail())
```

```python
print("Rows:", len(gender_clean))

print("\nMissing values:")
print(gender_clean.isnull().sum())

print("\nDuplicate years:",
      gender_clean["Year"].duplicated().sum())
```

Merging the two data sets:

```python
combined = pd.merge(
    population_clean,
    gender_clean,
    on="Year",
    how="inner"
)
print(combined)
print("Shape:", combined.shape)
```

Numerical Analysis

- Descriptive Statistics

```python
population_array = combined["Population"].to_numpy()

gender_ratio_array = combined[
    "Female_Male_Youth_Unemployment_Ratio"
].to_numpy()
```

```python
print(population_array)
print(gender_ratio_array)
```

- Minimum and Maximum

```python
print("POPULATION STATISTICS")
print("---------------------")
print("Mean:", np.mean(population_array))
print("Median:", np.median(population_array))
print("Minimum:", np.min(population_array))
print("Maximum:", np.max(population_array))
print("Standard Deviation:", np.std(population_array))
```

Population Statistics

The mean population of South Africa between 1991 and 2023 was approximately 51.51 million people, while the median population was approximately 50.53 million. The minimum population recorded in the dataset was 41,687,898 in 1991, while the maximum was 63,212,384 in 2023. This shows that South Africa's population increased considerably over the study period. The standard deviation was approximately 6.11 million people. This indicates that the population values varied by around 6.11 million people from the mean over the study period. Overall, the results show a clear upward trend in South Africa's population from 1991 to 2023.

```python
print("\nGENDER UNEMPLOYMENT RATIO STATISTICS")
print("------------------------------------")
print("Mean:", np.mean(gender_ratio_array))
print("Median:", np.median(gender_ratio_array))
print("Minimum:", np.min(gender_ratio_array))
print("Maximum:", np.max(gender_ratio_array))
print("Standard Deviation:", np.std(gender_ratio_array))
```

Gender Unemployment Ratio Statistics

The mean female-to-male youth unemployment ratio in South Africa between 1991 and 2023 was approximately 120.68, while the median was approximately 121.69. Since the mean and median are both above 100, the results indicate that the female youth unemployment rate was generally higher than the male youth unemployment rate during the study period. The lowest ratio was 113.69 in 2009, while the highest ratio was 126.08 in 2016. This means that the female-to-male youth unemployment ratio remained above 100 throughout the study period, although the size of the difference varied between years. The standard deviation was approximately 2.66, indicating relatively low variation in the ratio around its mean. Overall, the results suggest that the female-to-male youth unemployment relationship was relatively consistent between 1991 and 2023, despite some changes across individual years.

- Population Growth

```python

population_min_index = np.argmin(population_array)
population_max_index = np.argmax(population_array)

gender_min_index = np.argmin(gender_ratio_array)
gender_max_index = np.argmax(gender_ratio_array)

print("POPULATION")
print("Smallest population:", population_array[population_min_index])
print("Year:", combined["Year"].iloc[population_min_index])

print("Largest population:", population_array[population_max_index])
print("Year:", combined["Year"].iloc[population_max_index])

print("\nGENDER UNEMPLOYMENT RATIO")
print("Lowest ratio:", gender_ratio_array[gender_min_index])
print("Year:", combined["Year"].iloc[gender_min_index])

print("Highest ratio:", gender_ratio_array[gender_max_index])
print("Year:", combined["Year"].iloc[gender_max_index])
```

```python
population_growth = (
    (population_array[-1] - population_array[0])
    / population_array[0]
) * 100

print("Population growth from 1991 to 2023:",
      population_growth, "%")
```

Between 1991 and 2023, South Africa's total population increased by approximately 51.63%, indicating substantial population growth over the study period. The largest year-on-year increase occurred between 2014 and 2015, at approximately 2.03%.

```python
population_percent_change = (
    np.diff(population_array) / population_array[:-1]
) * 100

print(population_percent_change)
```

```python
max_growth_index = np.argmax(population_percent_change)

print("Largest annual population growth:",
      population_percent_change[max_growth_index], "%")

print("Year:",
      combined["Year"].iloc[max_growth_index + 1])
```

- Correlation

```python
correlation_matrix = np.corrcoef(
    population_array,
    gender_ratio_array
)

print(correlation_matrix)
```

```python
correlation = correlation_matrix[0, 1]

print("Correlation:", correlation)
```

The correlation coefficient between South Africa's total population and the female-to-male youth unemployment ratio from 1991 to 2023 was -0.301. This indicates a weak-to-moderate negative linear relationship. In general, as population increased, the female-to-male youth unemployment ratio tended to decrease. However, the relatively weak correlation means that population alone does not strongly explain changes in the unemployment ratio, and correlation does not imply causation

- Array reshaping

```python
population_reshaped = population_array.reshape(11, 3)

print("Original shape:", population_array.shape)
print("Reshaped shape:", population_reshaped.shape)

print("\nReshaped Population Array:")
print(population_reshaped)
```

The population array originally contained 33 yearly observations. Using NumPy's reshape() function, the one-dimensional array was transformed into an 11 × 3 two-dimensional array. This demonstrates how NumPy can restructure data into different dimensions while preserving the original values.

4. Database Integration

```python
import sqlite3
import pandas as pd

connection = sqlite3.connect("south_africa_analysis.db")

print("Database connected successfully!")
```

```python
combined.to_sql(
    "south_africa_data",
    connection,
    if_exists="replace",
    index=False
)

print("Table created successfully!")
```

```python
query = """
SELECT *
FROM south_africa_data
"""

database_data = pd.read_sql_query(query, connection)

print(database_data)
```

Demonstrate using an actual sql query

```python
query = """
SELECT Year, Population, Female_Male_Youth_Unemployment_Ratio
FROM south_africa_data
WHERE Population > 60000000
"""

result = pd.read_sql_query(query, connection)

print(result)
```

Safely update a record

```python
query = """
SELECT *
FROM south_africa_data
WHERE Year = 1991
"""

before_update = pd.read_sql_query(query, connection)

print("Before UPDATE:")
print(before_update)
```

```python
update_query = """
UPDATE south_africa_data
SET Population = 41687899
WHERE Year = 1991
"""

connection.execute(update_query)
connection.commit()

print("Record updated successfully!")
```

```python
query = """
SELECT *
FROM south_africa_data
WHERE Year = 1991
"""

after_update = pd.read_sql_query(query, connection)

print("After UPDATE:")
print(after_update)
```

```python
restore_query = """
UPDATE south_africa_data
SET Population = 41687898
WHERE Year = 1991
"""

connection.execute(restore_query)
connection.commit()

print("Original value restored.")
```

```python
query = """
SELECT *
FROM south_africa_data
WHERE Year = 1991
"""

print(pd.read_sql_query(query, connection))
```

5. Python data analysis

```python
analysis_df = combined.copy()

analysis_df["Population_Millions"] = (
    analysis_df["Population"] / 1_000_000
)

analysis_df["Population_Growth_%"] = (
    analysis_df["Population"].pct_change() * 100
)

analysis_df.head()
```

```python
summary_df = analysis_df[
    [
        "Year",
        "Population_Millions",
        "Population_Growth_%",
        "Female_Male_Youth_Unemployment_Ratio"
    ]
]

summary_df.head(10)
```

```python
def highlight_high(row):
    return [
        "background-color: yellow"
        if row["Population_Growth_%"] > 1.5
        else ""
        for _ in row
    ]

styled_summary = summary_df.style.apply(highlight_high, axis=1)

styled_summary
```

```python
highest_growth = analysis_df.loc[
    analysis_df["Population_Growth_%"].idxmax()
]

highest_ratio = analysis_df.loc[
    analysis_df["Female_Male_Youth_Unemployment_Ratio"].idxmax()
]

lowest_ratio = analysis_df.loc[
    analysis_df["Female_Male_Youth_Unemployment_Ratio"].idxmin()
]

print("Highest population growth:")
print(highest_growth)

print("\nHighest unemployment ratio:")
print(highest_ratio)

print("\nLowest unemployment ratio:")
print(lowest_ratio)
```

The female-to-male youth unemployment ratio reached its highest value of approximately 126.08 in 2016 and its lowest value of approximately 113.69 in 2009.
